```python
import subprocess

def get_video_title(url):
    result = subprocess.run(
        ["yt-dlp", "--get-title", url],
        capture_output=True,
        text=True,
        check=True
    )
    return result.stdout.strip()
```

```python
import re

def clean_filename(title):
    title = re.sub(r'[\\/*?:"<>|]', '', title)
    return title.strip()
```

```python
import os
import subprocess


def clip_and_save_with_progress(
    url,
    start,
    end,
    drive_folder="/content/drive/MyDrive/youtube_clips"
):
    os.makedirs(drive_folder, exist_ok=True)

    title = clean_filename(get_video_title(url))
    output_file = f"{drive_folder}/{title}_{start.replace(':','-')}_{end.replace(':','-')}.mp4"

    command = [
        "yt-dlp",
        "--newline",                       # 👈 enables live progress lines
        "--progress",                      # 👈 show progress
        "--download-sections", f"*{start}-{end}",
        "-f", "bv*+ba/b",
        "--merge-output-format", "mp4",
        "-o", output_file,
        url
    ]

    subprocess.run(command, check=True)
    return output_file
```

```python

clip_path = clip_and_save_with_progress(
    "https://www.youtube.com/watch?v=g5USCP4e6Cg",
    "00:26:40",
    "02:49:10"
)

print("Saved to Drive:", clip_path)
```
